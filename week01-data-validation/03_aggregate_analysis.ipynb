```python
# Show all rows for the group 'India' from the grouped object
country_grp = dataframe.groupby(['Country'])
country_grp.get_group('India')
```

```python
# For India, show counts of 'SocialMedia' values
filt = dataframe['Country'] == 'India'
dataframe.loc[filt]['SocialMedia'].value_counts()
```

```python
# Median 'ConvertedComp' specifically for India
country_grp['ConvertedComp'].median().loc['India']
```

```python
# Count Python users in India by checking 'LanguageWorkedWith' contains 'Python'
filt = dataframe['Country'] == 'India'
dataframe.loc[filt]['LanguageWorkedWith'].str.contains('Python').sum()
```

```python
# Properly count Python mentions per country using apply on grouped series
country_grp['LanguageWorkedWith'].apply(lambda x: x.str.contains('Python').sum())
```

```python
# Create a series with counts of Python users per country
Country_respondents = dataframe['Country'].value_counts()
Country_respondents
```

```python
# Count how many respondents per country list Python in 'LanguageWorkedWith'
country_uses_python = country_grp['LanguageWorkedWith'].apply(lambda x: x.str.contains('Python').sum())
country_uses_python
```

```python
# Combine the two series into a DataFrame
python_df = pd.concat([Country_respondents, country_uses_python], axis='columns', sort=False)
python_df
```

```python
# renaming columns for clarity
python_df.rename(columns={'count':'NumRespondents', 'LanguageWorkedWith':'NumUsingPython'}, inplace=True)
python_df
```

```python
# Compute percentage of respondents per country who know Python
python_df['%knowsPython'] = (python_df['NumUsingPython'] / python_df['NumRespondents']) * 100
python_df
```

```python
# sort which country has highest percentage of Python users
python_df.sort_values(by='%knowsPython', ascending=False, inplace=True)
python_df.head(50)
```
