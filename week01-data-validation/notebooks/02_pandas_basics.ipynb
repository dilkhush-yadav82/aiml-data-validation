Aggregate Statistics (Group By)

```python
df3 = pd.read_csv('RAW_DATA\\pokemon_data.csv')
```

```python
df3
```

```python
# Groupby 'Type 1' and get the mean of each numeric column
df3.groupby(['Type 1']).mean(numeric_only=True)
```

```python
# group by 'Type 1' and sort by multiple columns
# Here we are sorting by 'HP', 'Attack', and 'Defense' in descending order
df3.groupby(['Type 1']).mean(numeric_only=True).sort_values(['HP', 'Attack', 'Defense'], ascending=False)
```

```python
# sum of each numeric column grouped by 'Type 1'
df3.groupby(['Type 1']).sum(numeric_only=True)
```

```python
# count number of pokemon by 'Type 1' and get counts of all columns 
df3.groupby(['Type 1']).count()
```

```python
df3['count'] = 1
df3
```

```python

df3.groupby(['Type 1', 'Type 2']).count() ['count']
```

working with Large datasets

```python
new_df = pd.DataFrame(columns=df3.columns)
for df3 in pd.read_csv('RAW_DATA\\pokemon_data.csv', chunksize=5):
    results = df3.groupby(['Type 1']).count()
    new_df = pd.concat([new_df, results])   
```
