```python
# Concept: Import NumPy for numeric operations and represent missing values with `np.nan`
# Remember: Use `np.nan` consistently when preparing data for pandas
import numpy as np
```

```python
import pandas as pd
```

```python
# Concept: Create a small dataset demonstrating different missing-value markers
# Remember: Strings like 'NA' or 'Missing' are not np.nan until replaced
people = {
    'first': ['Corey', 'Jane', 'John', 'Chris', np.nan, None, 'NA'],
    'last': ['Schafer', 'Doe', 'Doe', 'Schafer', np.nan, np.nan, 'Missing'],
    'email': [
        'CoreyMSchafer@gmail.com',
        'JaneDoe@email.com',
        'JohnDoe@email.com', None, np.nan, 'Anonymous@email.com','NA'],
    'age': ['33', '55', '63', '36', None, None, 'Missing']
}
```

```python
# Concept: Build a DataFrame from the example dictionary to inspect missing data
# Remember: Displaying the DataFrame helps decide cleaning strategy
df=pd.DataFrame(people)
df
```

```python
# Concept: `dropna()` removes rows with missing values by default
# Remember: Default drops any row with at least one NA (how='any')
df.dropna()
```

```python
# Concept: Explicit `axis='index'` with `how='any'` drops rows having any NA
# Remember: This is equivalent to the default `dropna()` behavior
df.dropna(axis='index', how='any')
```

```python
# Concept: `how='all'` drops rows only when all values are missing
# Remember: Use when rows with at least one valid value should be kept
df.dropna(axis='index', how='all')
```

```python
# Concept: Drop columns with all missing values using `axis='columns'`
# Remember: Column-wise drops change DataFrame shape significantly
df.dropna(axis='columns', how='all')
```

```python
# Concept: Drop columns containing any missing value with `how='any'`
# Remember: This is aggressive — often not what you want for real data
df.dropna(axis='columns', how='any')
```

```python
# Concept: Use `subset` to limit `dropna()` to specific columns
# Remember: Only rows missing in the subset columns are considered for dropping
df.dropna(axis='index', how='any', subset=['email'])
```

```python
# Concept: `how='all'` with subset keeps rows that have any value in the subset
# Remember: Use this when you only want to drop rows that lack all subset values
df.dropna(axis='index', how='all', subset=['email'])
```

```python
# Concept: Combine multiple columns in `subset` to control dropping behavior
# Remember: `how='all'` vs `how='any'` yields different retention rules
df.dropna(axis='index', how='all', subset=['last','email'])
```

```python
# Concept: Use `how='any'` with multiple subset columns to drop rows missing any of them
# Remember: This removes rows even if only one of the subset columns is missing
df.dropna(axis='index', how='any', subset=['last','email'])
```

```python
# Concept: Replace sentinel strings with `np.nan` to standardize missing values
# Remember: `inplace=True` mutates the DataFrame; keep originals if needed
df2 = pd.DataFrame(people)
df2.replace('NA', np.nan, inplace=True)
df2.replace('Missing', np.nan, inplace=True)
```

```python
# Concept: Inspect the DataFrame after replacing sentinel values
# Remember: Confirm replacements converted string markers to actual NaN
df2
```

```python
# Concept: Drop rows missing 'email' after sentinel replacements
# Remember: Work on a copy if you need the original data preserved
df2.dropna(axis='index', how='any', subset=['email'])
```

```python
# Concept: Drop rows only when the subset columns are all missing
# Remember: This keeps rows that have at least one value in the subset
df2.dropna(axis='index', how='all', subset=['email'])
```

```python
# Concept: Use subset with multiple columns to control dropping behavior
# Remember: Choose 'all' when you only want to drop rows missing every listed column
df2.dropna(axis='index', how='all', subset=['last','email'])
```

```python
# Concept: Use 'any' to drop rows missing any of the subset columns
# Remember: This is stricter — rows missing one of the subset columns will be removed
df2.dropna(axis='index', how='any', subset=['last','email'])
```

```python
# Concept: `isna()` returns a boolean mask indicating missing entries
# Remember: Use the mask for conditional operations or to count missing values
df2.isna()
```

```python
# Concept: `fillna()` provides a simple way to replace NaNs with a value
# Remember: Prefer domain-appropriate fills; avoid misleading default values
df2.fillna('HELLOFILL')
```

```python
# Concept: Check column data types to plan conversions or analyses
# Remember: Non-numeric strings prevent numeric aggregation like mean
df2.dtypes
```

```python
# Concept: Computing mean fails on non-numeric types — requires conversion
# Remember: Convert or coerce types before aggregation to avoid errors
df2['age'].mean()
```

```python
# Concept: Use `astype()` to cast a Series to an integer type
# Remember: Casting strings with non-numeric values will raise errors
df2['age'].astype('int')
```

```python
# Concept: Convert 'age' to float first to accommodate NaNs or decimal values
# Remember: Use `errors='coerce'` if you expect invalid strings you want as NaN
df2['age'] = df2['age'].astype('float')
```

```python
# Concept: Re-check dtypes after casting to confirm conversions succeeded
# Remember: Downstream ops (like mean) depend on correct numeric types
df2.dtypes
```

```python
# Concept: Now that 'age' is numeric, compute its mean
# Remember: Results can be influenced by NaNs — consider `.mean(skipna=True)` or filling missing values first
df2['age'].mean()
```
