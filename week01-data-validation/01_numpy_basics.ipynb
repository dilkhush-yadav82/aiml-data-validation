Linear algebra

````python
a1=np.ones((2,3))
a1
````

````python
a2=np.full((3,2), 5)
a2
````

````python
# find multiplicatuion of matrix a*b with b*a   
np.matmul(a1,a2)
````

````python
# determinant of a matrix
b1 =np.eye(6)
np.linalg.det(b1)
````

````python
# Trace = sum of diagonal elements of a square matrix
np.trace(b1)
````

````python

A = np.array([[1, 2],
              [3, 4]])

U, S, VT = np.linalg.svd(A)
print("U Matrix:\n", U)
print("Singular Values:\n", S)  
print("VT Matrix:\n", VT)   
````

````python

# Inverse
A_inv = np.linalg.inv(A)
print("Inverse of A:\n", A_inv)
````

````python
# Eigenvalues and Eigenvectors
B = np.array([[2, 0],
              [0, 3]])

values, vectors = np.linalg.eig(B)
print("Eigenvalues:\n", values)
print("Eigenvectors:\n", vectors)
````

````python
#matrix norm = measure of matrix's magnitude
np.linalg.norm(B)
````

````python
#vertical stacking
v1 = np.array([1,2,3,4])
v2 = np.array([5,6,7,8]) 
np.vstack([v1,v2])
````

````python
#vertical stacking
v11 = np.ones((2,4))
v22 = np.zeros((2,4))
np.vstack([v11,v22])
````

````python
#horizontal stacking
h1 = np.array([1,2,3,4])
h2 = np.array([5,6,7,8])
np.hstack([h1,h2])
````

````python
#horizontal stacking
h11 = np.ones((2,4))
h22 = np.zeros((2,4))
np.hstack([h11,h22])
````
